### Encoding is necessary to convert categorical value into numeric value so we can perform calculation

Note: Don't run the cell twice otherwise the code will encode all the values to 0

```python
# Enconding data in dataset
def convert(value):        #function to convert Yes and NO to 1 and 0
    if value == 'Yes':
        return 1
    else:
        return 0

X['Dependents'] = X['Dependents'].apply(convert)         #apply changes to the dataset X
X['PhoneService']= X['PhoneService'].apply(convert)
X['MultipleLines']= X['MultipleLines'].apply(convert)
X.head()
        
```

## Feature scaling (normalization)

Numerical columns need scaling because larger numbers like  tenure, monthlyCharges don't disproportionately dominate smaller numbers like tenure or binary flags).

Categorical columns need encoding because machine learning algorithmsrequire numbers instead of text strings like "Fiber optic" or "Month-to-month").

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_train
```

### Purpose of Saving the instance of scaler
a. When you .fit_transform() your training data, the scaler learns specific parameters—namely, the mean and standard deviation of each column in your training set.

b.When new, unseen customer data comes in later (or when you evaluate your model on your test set), that new data must be scaled using the exact same mean and standard deviation learned from the training data.

c. Saving the scaler ensures you can load it later using joblib.load('scaler.pkl') to transform new inputs identically.


```python
import joblib
joblib.dump(scaler, "scaler.pkl")     # saves the scaler.pkl file in current working directory
```
